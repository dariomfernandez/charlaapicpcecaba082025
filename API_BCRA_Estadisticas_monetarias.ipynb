```python
import requests
from requests.packages import urllib3
import json
import pandas as pd

def json_bonito(json_feo):
  return(json.dumps(json_feo, indent=4))

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

maestro_monetarias = requests.get("https://api.bcra.gob.ar/estadisticas/v3.0/Monetarias", verify=False)

print(json_bonito(maestro_monetarias.json()))
```

```python
def json_bonito(json_feo):
  return(json.dumps(json_feo, indent=4))

monetarias_param = {"desde": "2025-01-01",
                    "hasta": "2025-05-21"}

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

maestro_monetarias = requests.get("https://api.bcra.gob.ar/estadisticas/v3.0/Monetarias/1", verify=False, params=monetarias_param)

print(json_bonito(maestro_monetarias.json()))
```
