```python
# Vamos a obtener un listado de CUITs para consultar sus deudas
# desde una hoja de Excel que tengo en mi propio Drive
# en el path "/content/drive/MyDrive/Colab Notebooks/Lista bancos.xlsx"
df_bancos = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/CharlaAPICPCECABA2025/Lista bancos.xlsx')
# Imprimo las primeras 5 filas (con el método head), para ver si importé bien la hoja
print(df_bancos.head())
```
